```python
# COMMAND ----------

dbutils.secrets.list('salesproject')
```

```python
# COMMAND ----------

clientid = dbutils.secrets.get("salesproject", "clientid")
clientsecret = dbutils.secrets.get("salesproject", "secretvalue")
tenantid = dbutils.secrets.get("salesproject", "tenantid")
```

```python
# COMMAND ----------

storage_account = "salesstorageproject"

spark.conf.set(
    f"fs.azure.account.auth.type.{storage_account}.dfs.core.windows.net",
    "OAuth"
)

spark.conf.set(
    f"fs.azure.account.oauth.provider.type.{storage_account}.dfs.core.windows.net",
    "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider"
)

spark.conf.set(
    f"fs.azure.account.oauth2.client.id.{storage_account}.dfs.core.windows.net",
    dbutils.secrets.get("salesproject", "clientid")
)

spark.conf.set(
    f"fs.azure.account.oauth2.client.secret.{storage_account}.dfs.core.windows.net",
    dbutils.secrets.get("salesproject", "secretvalue")
)

spark.conf.set(
    f"fs.azure.account.oauth2.client.endpoint.{storage_account}.dfs.core.windows.net",
    f"https://login.microsoftonline.com/{dbutils.secrets.get('salesproject', 'tenantid')}/oauth2/token"
)
```
