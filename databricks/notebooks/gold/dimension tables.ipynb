```python
customer = spark.table("sales.silver.customer")

dim_customer = customer.select(
    "CustomerID",
    "FirstName",
    "LastName",
    "City",
    "Email"
)

dim_customer.write.mode("append").saveAsTable("sales.gold.dim_customer")
```

```python
products = spark.table("sales.silver.products")

dim_product = products.select(
    "ProductID",
    "ProductName",
    "Category",
    "Price"
)

dim_product.write.mode("append").saveAsTable("sales.gold.dim_product")
```

```python
stores = spark.table("sales.silver.stores")

dim_store = stores.select(
    "StoreID",
    "StoreName",
    "City",
    "Province",
    "region"
)

dim_store.write.mode("append").saveAsTable("sales.gold.dim_store")
```
